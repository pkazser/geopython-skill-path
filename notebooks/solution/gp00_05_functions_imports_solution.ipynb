```python
def m2_to_ha(area_m2):
    return area_m2 / 10000

print(m2_to_ha(25000))
```

```python
def rectangle_area(width, height):
    return width * height

print(rectangle_area(100, 50))
```

```python
def elevation_class(elevation):
    if elevation < 100:
        return "Low"
    elif elevation < 500:
        return "Medium"
    else:
        return "High"

print(elevation_class(80))
print(elevation_class(250))
print(elevation_class(750))
```

```python
from shapely.geometry import Point

stations = [
    {"name": "S1", "x": 23.55, "y": 41.09},
    {"name": "S2", "x": 23.56, "y": 41.10},
    {"name": "S3", "x": 23.57, "y": 41.08}
]

def make_point(x, y):
    return Point(x, y)

geometries = []

for station in stations:
    geometry = make_point(station["x"], station["y"])
    geometries.append(geometry)

print(geometries)
```
