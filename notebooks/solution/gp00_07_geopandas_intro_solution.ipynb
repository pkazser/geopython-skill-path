# GP00.07 — Πρώτη επαφή με GeoPandas

## Solution Notebook

```python
import pandas as pd

data = {
    "name": ["P1", "P2", "P3"],
    "longitude": [23.55, 23.56, 23.57],
    "latitude": [41.09, 41.10, 41.08]
}

df = pd.DataFrame(data)
df
```

```python
import geopandas as gpd

gdf = gpd.GeoDataFrame(
    df,
    geometry=gpd.points_from_xy(
        df["longitude"],
        df["latitude"]
    ),
    crs="EPSG:4326"
)

gdf
```

```python
print("CRS:", gdf.crs)
print("Rows:", len(gdf))
print("Columns:", list(gdf.columns))
print("Bounds:", gdf.total_bounds)
print("Geometry types:")
print(gdf.geometry.geom_type.value_counts())
```

```python
print("Missing geometries:", gdf.geometry.isna().sum())
print("Empty geometries:", gdf.geometry.is_empty.sum())
```

```python
gdf.plot(figsize=(6, 6))
```

```python
gdf.to_file("points.geojson", driver="GeoJSON")
```

```python
import pandas as pd
import geopandas as gpd

data = {
    "name": ["A", "B", "C", "D"],
    "longitude": [23.55, 23.56, 23.57, 23.58],
    "latitude": [41.09, 41.10, 41.08, 41.11],
    "elevation": [52.3, 65.1, 48.9, 71.4]
}

df = pd.DataFrame(data)

gdf = gpd.GeoDataFrame(
    df,
    geometry=gpd.points_from_xy(
        df["longitude"],
        df["latitude"]
    ),
    crs="EPSG:4326"
)

print(gdf)
print("CRS:", gdf.crs)
print("Rows:", len(gdf))
print("Geometry types:")
print(gdf.geometry.geom_type.value_counts())
print("Bounds:")
print(gdf.total_bounds)

gdf.plot(figsize=(6, 6))
```
