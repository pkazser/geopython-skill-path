```python
points = [
 {"name":"P1","elevation":82,"active":True},
 {"name":"P2","elevation":125,"active":True},
 {"name":"P3","elevation":150,"active":False},
 {"name":"P4","elevation":110,"active":True},
 {"name":"P5","elevation":65,"active":False}
]
selected_points = []
for point in points:
    if point["active"] and point["elevation"] > 100:
        selected_points.append(point)
print(selected_points)
print("Selected points:", len(selected_points))
```
