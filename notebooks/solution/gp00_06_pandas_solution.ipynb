```python
import pandas as pd

data = {
    "city": ["Serres", "Drama", "Kavala"],
    "population": [60000, 45000, 70000]
}

df = pd.DataFrame(data)
df
```

```python
print(df.head())
print(df.shape)
print(df.columns)
print(df.dtypes)
```

```python
data = {
    "parcel": ["P1", "P2", "P3"],
    "area_m2": [5000, 12500, 30000]
}

df = pd.DataFrame(data)
df["area_ha"] = df["area_m2"] / 10000
df
```

```python
data = {
    "name": ["A", "B", "C", "D"],
    "elevation": [80, 125, 95, 160],
    "active": [True, True, False, True]
}

df = pd.DataFrame(data)

selected = df[
    df["active"] &
    (df["elevation"] > 100)
]

selected
```

```python
data = {
    "name": ["P1", "P2", "P3"],
    "elevation": [80, None, 120]
}

df = pd.DataFrame(data)
print(df.isna().sum())
```

```python
data = {
    "city": ["Serres", "Drama", "Kavala", "Thessaloniki"],
    "population": [60000, 45000, 70000, 320000]
}

df = pd.DataFrame(data)

sorted_df = df.sort_values("population", ascending=False)
sorted_df
```

```python
data = {
    "id": [1, 2, 3, 4, 5],
    "name": ["P1", "P2", "P3", "P4", "P5"],
    "elevation": [82, 125, 95, 160, 110],
    "active": [True, True, False, True, True]
}

df = pd.DataFrame(data)

print(df.head())
print(df.shape)
print(df.dtypes)
print("Mean elevation:", df["elevation"].mean())

selected = df[
    df["active"] &
    (df["elevation"] > 100)
]

selected
```
