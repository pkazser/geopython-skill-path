```python
import pandas as pd
import geopandas as gpd

data = {
    "name": ["P1","P2","P3"],
    "longitude": [23.55,23.56,23.57],
    "latitude": [41.09,41.10,41.08]
}

df = pd.DataFrame(data)

gdf = gpd.GeoDataFrame(
    df,
    geometry=gpd.points_from_xy(
        df["longitude"],
        df["latitude"]
    ),
    crs="EPSG:4326"
)

print("CRS:", gdf.crs)
print("Rows:", len(gdf))
print("Bounds:", gdf.total_bounds)
gdf.plot()
```

## Reflection — ενδεικτικές απαντήσεις

1. Ένα DataFrame έχει attributes αλλά όχι ειδική geometry και CRS.
2. Το GeoDataFrame προσθέτει geometry και χωρικές δυνατότητες.
3. CRS και bounds βοηθούν να ελέγξουμε θέση, μονάδες και λογική συνέπεια.
4. Ο κώδικας μπορεί να εκτελείται κανονικά αλλά να έχει γεωχωρικά λανθασμένη λογική.
