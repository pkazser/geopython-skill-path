```python
import pandas as pd
import geopandas as gpd

data = {
    "name": ["P1","P2","P3","P4","P5","P6"],
    "category": ["school","hospital","school","clinic","library","hospital"],
    "longitude": [23.550,23.556,23.562,23.568,23.574,23.580],
    "latitude": [41.090,41.094,41.087,41.101,41.096,41.084],
    "elevation": [82,125,95,160,110,72],
    "active": [True,True,False,True,True,True]
}

df = pd.DataFrame(data)

print(df.head())
print(df.shape)
print(df.columns)
print(df.dtypes)

gdf = gpd.GeoDataFrame(
    df,
    geometry=gpd.points_from_xy(df["longitude"], df["latitude"]),
    crs="EPSG:4326"
)

print("CRS:", gdf.crs)
print("Rows:", len(gdf))
print(gdf.geometry.geom_type.value_counts())
print("Bounds:", gdf.total_bounds)
print("Missing geometries:", gdf.geometry.isna().sum())

gdf.plot(figsize=(6, 6))
```

```python
selected = gdf[
    gdf["active"] &
    (gdf["elevation"] > 100)
].copy()

print(selected[["name", "category", "elevation"]])
```

```python
active_hospitals = gdf[
    gdf["active"] &
    (gdf["category"] == "hospital")
].copy()

print(active_hospitals[["name", "category"]])
```

```python
def elevation_class(elevation):
    if elevation < 100:
        return "Low"
    elif elevation < 150:
        return "Medium"
    else:
        return "High"

gdf["elevation_class"] = gdf["elevation"].apply(elevation_class)

print(gdf[["name", "elevation", "elevation_class"]])
```

```python
gdf.to_file(
    "gp00_points.gpkg",
    layer="points",
    driver="GPKG"
)

selected.to_file(
    "gp00_selected.geojson",
    driver="GeoJSON"
)
```

```python
priority_points = gdf[
    gdf["active"]
    &
    (
        (gdf["category"] == "school")
        |
        (gdf["category"] == "hospital")
    )
    &
    (gdf["elevation"] >= 90)
].copy()

print(priority_points[["name", "category", "elevation"]])
```
