## 10. `GeoPandas`


```python
import geopandas as gpd
from shapely.geometry import Point

data = {"name": ["A", "B", "C"], "longitude": [23.55, 23.56, 23.57], "latitude": [41.09, 41.10, 41.08]}
geometry = [Point(xy) for xy in zip(data["longitude"], data["latitude"])]
gdf = gpd.GeoDataFrame(data, geometry=geometry, crs="EPSG:4326")
print("CRS:", gdf.crs)
print("Geometry types:")
print(gdf.geometry.geom_type)
print("Rows:", len(gdf))
gdf
```

## Mini Challenge — Ενδεικτική λύση


```python
import geopandas as gpd
from shapely.geometry import Point

data = {"name": ["P1", "P2", "P3", "P4"], "longitude": [23.55, 23.56, 23.57, 23.58], "latitude": [41.09, 41.10, 41.08, 41.11], "elevation": [52.3, 65.1, 48.9, 71.4]}
geometry = [Point(xy) for xy in zip(data["longitude"], data["latitude"])]
gdf = gpd.GeoDataFrame(data, geometry=geometry, crs="EPSG:4326")
print("CRS:", gdf.crs)
print("Rows:", len(gdf))
gdf
```
